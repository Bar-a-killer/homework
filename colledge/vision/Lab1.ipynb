題目:撰寫利用 Python 檢查輸入數字是否是質數的程式

```python
def is_prime(n: int) -> bool:
    if n < 2:
        return False
    if n % 2 == 0:
        return n == 2
    if n % 3 == 0:
        return n == 3
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True

i = 0;
i = int(input());
if is_prime(i):
  print("Is prime")
else:
  print("Not prime")
```

作法與方法迭帶

直覺的方法：從 2 檢查到 n-1，看看有沒有能整除 n 的數。

問題:效率差，O(n)。

改良 1:只檢查到 √n

改良 2:跳過明顯不可能的數 如2、3

改良 3:6k ± 1 法則

所有質數(除了 2, 3)都可以寫成 6k±1 的形式。

解釋:

整數除以 6，餘數可能是 0, 1, 2, 3, 4, 5。

若餘數 = 0, 2, 4 → 偶數，不可能是質數(除了 2)。

若餘數 = 3 → 能被 3 整除，不可能是質數(除了 3)。

剩下只有餘數 1, 5(即 6k+1, 6k-1)。

→ 只需要測試 5, 7, 11, 13, 17, …(每次加 6，測試 i 與 i+2)。

心得:

這次練習主要是撰寫一個判斷質數的程式。程式中我採用了 6k±1 的方式 來加速判斷，這比傳統從 2 開始一個一個測試的方式更有效率，因為除了 2 和 3 以外，所有質數都可以表示成 6k±1 的形式。這樣能把需要檢查的數量大幅減少，提高效能。
